```python
import pyodbc 
import pandas as pd
```

```python
pyodbc.drivers()  # List available ODBC drivers
```

```python
# Establish connection to the source database
source_conn = pyodbc.connect(
    "DRIVER={ODBC Driver 17 for SQL Server};"
    "SERVER=.\SQLEXPRESS;"
    "DATABASE=Sales_OLTP;"
    "Trusted_Connection=yes;"
    )
```

```python
# Working On DimCustomer table
customer_df = pd.read_sql("select * from Customer", source_conn)
customer_df
```

```python
# We need also Gender Name, so we will use left outer join,
# we need CustomerID-->PK, FullName, PhoneNumber, Email, GenderName
customer_df = pd.read_sql("select CustomerID," \
" CONCAT(FirstName ,' ' ,LastName) as FullName,PhoneNumber as Phone," \
" Email, GenderName from Customer c " \
"left outer join Gender g " \
"on c.GenderID = g.GenderID", source_conn)
customer_df
```

```python
# Working On DimProduct table ProductID --> PK, ProductName, Price, SubCategoryName, CategoryName
product_df = pd.read_sql("select ProductID, ProductName, Price, SubCategoryName,CategoryName" \
" from prod_view p left outer join Category c " \
"on p.CategoryID = c.CategoryID ",source_conn)
product_df
```

```python
# Working On DimSalesMan table SalesManID --> PK, FullName, Phone, Email, City
salesman_df = pd.read_sql("select * from salesman_view", source_conn)
salesman_df
```

```python
# Working on FactOrders table
factorders_df = pd.read_sql("select * from order_view", source_conn)
factorders_df
```

```python
# order_date is date not int so we need to change it to int to be able to load it in the data warehouse
# also convert it from 2023-01-01 10:00:00 to 20230101
factorders_df['Order_Date'] = factorders_df['Order_Date'].dt.strftime('%Y%m%d').astype(int)
factorders_df.head()
```

## 3 - Load

```python
# Establish connection to the Destination database TO DWH
destination_conn = pyodbc.connect(
    "DRIVER={ODBC Driver 17 for SQL Server};"
    "SERVER=.\SQLEXPRESS;"
    "DATABASE=Sales_DWH;"
    "Trusted_Connection=yes;"
    )
```

```python
for index, row in customer_df.iterrows():
    cursor = destination_conn.cursor()
    cursor.execute("INSERT INTO DimCustomer (CustomerID, FullName, Phone, Email, GenderName) VALUES (?, ?, ?, ?, ?)",
                   row['CustomerID'], row['FullName'], row['Phone'], row['Email'], row['GenderName'])
    destination_conn.commit()
```

```python
for index, row in product_df.iterrows():
    cursor = destination_conn.cursor()
    cursor.execute("INSERT INTO DimProduct (ProductID, ProductName, Price, SubCategoryName, CategoryName) VALUES (?, ?, ?, ?, ?)",
                   row['ProductID'], row['ProductName'], row['Price'], row['SubCategoryName'], row['CategoryName'])
    destination_conn.commit()
```

```python
for index, row in salesman_df.iterrows():
    cursor = destination_conn.cursor()
    cursor.execute("INSERT INTO DimSalesMan (SalesManID, FullName, Phone, Email, City) VALUES (?, ?, ?, ?, ?)",
                   row['SalesManID'], row['FullName'], row['Phone'], row['Email'], row['City'])
    destination_conn.commit()
```

```python
for index, row in factorders_df.iterrows():
    cursor = destination_conn.cursor()
    cursor.execute("INSERT INTO FactOrders (OrderID, ProductID, CustomerID, SalesManID, Order_Date, Quantity,TotalPrice) VALUES (?, ?, ?, ?, ?, ?, ?)",
                   row['OrderID'], row['ProductID'], row['CustomerID'], row['SalesManID'], row['Order_Date'], row['Quantity'], row['TotalPrice'])
    destination_conn.commit()
```

```python
cursor.close()
destination_conn.close()
```
